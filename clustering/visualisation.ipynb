## Cluster Visualisation

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE, trustworthiness
import plotly.express as px
import matplotlib.pyplot as plt
import scipy.stats as sp
```

```python
import os
os.chdir(r"C:\Users\belle\OneDrive\Personal Projects\athlete_profiles\athlete_clustering")
```

```python
df = pd.read_csv("data/athlete_summary.csv")
df.head(3).T
```

```python
scaled_data = StandardScaler().fit_transform(df.drop(['Athlete_ID','Sport'],axis=1).select_dtypes(include=[np.number]))
```

```python
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)
```

```python
fig = px.scatter(x=pca_result[:, 0], 
                 y=pca_result[:, 1], 
                 labels={'x': 'PCA 1', 'y': 'PCA 2'}, 
                 title='PCA of Athlete Data',
                 height=800, width=800)
fig.show()
```

```python
tsne = TSNE(n_components=2, perplexity=5, random_state=59)
tsne_result = tsne.fit_transform(scaled_data)
print(f"Trustworthiness of t-SNE embedding: {trustworthiness(scaled_data, tsne_result)}")
```

```python
fig = px.scatter(x=tsne_result[:, 0],
                 y=tsne_result[:, 1],  
                 labels={'x': 't-SNE 1', 'y': 't-SNE 2'}, 
                 title='t-SNE of Athlete Data',
                 height=800, width=800)
fig.show()
```
